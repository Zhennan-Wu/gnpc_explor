```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set visualization style for all plots
sns.set_theme(style="whitegrid", palette="muted")

# Define the directory where your 17 CSV files are stored
DATA_DIR = '.' 

# ---------------------------------------------------------
# 1. DATA LOADING
# ---------------------------------------------------------
def load_data(filename):
    """Safely load a CSV file into a pandas DataFrame."""
    file_path = os.path.join(DATA_DIR, filename)
    try:
        print(f"Loading {filename}...")
        return pd.read_csv(file_path, low_memory=False)
    except FileNotFoundError:
        print(f"  -> Warning: {filename} not found. Skipping.")
        return None

print("=== PART 1: LOADING DATASETS ===")
datasets = {
    's2_shift_cov_lap': load_data('results_S2_model_shift_mean_cov_lap.csv'),
    's3_fix_na_lap': load_data('results_S3_model_fix_mean_lap.csv'),
    's3_fix_na_na': load_data('results_S3_model_fix_mean.csv'),
    's5_shift_global_lap': load_data('results_S5_model_shift_mean_paper_lap.csv'),
    's5_shift_global_na': load_data('results_S5_model_shift_mean_paper.csv')
}
```

```python
# ---------------------------------------------------------
# 2. MERGE DATASETS BY SCENARIO
# ---------------------------------------------------------

def merge_by_scenario(datasets):
    scenario_groups = {}

    # Step 1: group datasets by scenario
    for name, df in datasets.items():
        if df is None:
            continue
        
        scenario = name.split('_')[0]   # s2, s3, s5
        
        if scenario not in scenario_groups:
            scenario_groups[scenario] = []
        
        # rename columns except Parameter
        df = df.copy()
        rename_dict = {
            col: f"{col}_{name}" for col in df.columns if col != "Parameter"
        }
        df = df.rename(columns=rename_dict)
        
        scenario_groups[scenario].append(df)

    # Step 2: merge within each scenario
    merged_results = {}
    
    for scenario, dfs in scenario_groups.items():
        merged_df = dfs[0]
        
        for df in dfs[1:]:
            merged_df = merged_df.merge(df, on="Parameter", how="outer")
        
        merged_results[scenario] = merged_df

    return merged_results


merged_scenarios = merge_by_scenario(datasets)

print("\n=== MERGED DATASETS ===")
for k, v in merged_scenarios.items():
    print(f"{k}: {v.shape}")
```

```python
from collections import defaultdict

scenario_groups = defaultdict(dict)

for name, df in datasets.items():
    if df is None:
        continue
    
    scenario = name.split('_')[0]   # S2, S3, S5
    model = "_".join(name.split('_')[1:])  # model type
    
    scenario_groups[scenario][model] = df
```

```python
merged_results = {}

for scenario, models in scenario_groups.items():
    
    merged_df = None
    
    for model_name, df in models.items():
        
        df_model = df.copy()
        
        # Rename columns except Parameter
        rename_cols = {
            c: f"{model_name}_{c}" 
            for c in df_model.columns if c != "Parameter"
        }
        
        df_model = df_model.rename(columns=rename_cols)
        
        if merged_df is None:
            merged_df = df_model
        else:
            merged_df = pd.merge(
                merged_df,
                df_model,
                on="Parameter",
                how="outer"
            )
    
    merged_results[scenario] = merged_df
```

```python
def reorder_columns(df):
    
    param_col = ['Parameter']
    other_cols = [c for c in df.columns if c != 'Parameter']
    
    metrics = sorted(set(c.split('_')[-1] for c in other_cols))
    
    ordered = []
    
    for metric in metrics:
        metric_cols = [c for c in other_cols if c.endswith(metric)]
        ordered.extend(sorted(metric_cols))
    
    return df[param_col + ordered]

for scenario in merged_results:
    merged_results[scenario] = reorder_columns(merged_results[scenario])
```
