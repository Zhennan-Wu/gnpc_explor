```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from cmdstanpy import CmdStanModel

# ---------------------------------------------------------
# 1. DATA PREPARATION (ALSFRS-R Subset)
# ---------------------------------------------------------
print("Loading ALSFRS data...")
DATA_DIR = '../data/PRO-ACT_Data/2026_02_27_PROACT_ALL_FORMS/' 
df_als = pd.read_csv(DATA_DIR + 'F_PROACT_ALSFRS.csv', low_memory=False)

# The 12 items for ALSFRS-Revised
items_cols = [
    'Q1_Speech', 'Q2_Salivation', 'Q3_Swallowing',                 # Factor 1: Bulbar
    'Q4_Handwriting', 'Q5_Cutting', 'Q6_Dressing_and_Hygiene',     # Factor 2: Fine Motor
    'Q7_Turning_in_Bed', 'Q8_Walking', 'Q9_Climbing_Stairs',       # Factor 3: Gross Motor
    'R_1_Dyspnea', 'R_2_Orthopnea', 'R_3_Respiratory_Insufficiency'# Factor 4: Respiratory
]

# Merge Q5a and Q5b into a single feature
df_als['Q5a_Cutting_without_Gastrostomy'] = pd.to_numeric(df_als['Q5a_Cutting_without_Gastrostomy'], errors='coerce')
df_als['Q5b_Cutting_with_Gastrostomy'] = pd.to_numeric(df_als['Q5b_Cutting_with_Gastrostomy'], errors='coerce')
df_als['Q5_Cutting'] = df_als['Q5a_Cutting_without_Gastrostomy'].fillna(df_als['Q5b_Cutting_with_Gastrostomy'])

for col in items_cols + ['ALSFRS_Delta']:
    if col in df_als.columns:
        df_als[col] = pd.to_numeric(df_als[col], errors='coerce')

# Drop missing targets and strictly sort by time
df_als = df_als.dropna(subset=['ALSFRS_Delta', 'R_1_Dyspnea'])
df_als = df_als.sort_values(by=['subject_id', 'ALSFRS_Delta'])

# CRITICAL FIX: Drop same-day duplicate rows to prevent Delta T = 0 covariance crashes!
df_als = df_als.drop_duplicates(subset=['subject_id', 'ALSFRS_Delta'], keep='first')

# Sample ~100 patients to ensure compilation and modeling runs in a reasonable timeframe. 
# (You can increase this to the full dataset once testing is complete)
np.random.seed(42)
sampled_subjects = np.random.choice(df_als['subject_id'].unique(), size=100, replace=False)
df_subset = df_als[df_als['subject_id'].isin(sampled_subjects)].copy()

# Remap IDs sequentially for Stan (1 to N_sub)
df_subset['stan_subj_id'] = df_subset.groupby('subject_id').ngroup() + 1

# Calculate Delta T
df_subset['delta_t'] = df_subset.groupby('subject_id')['ALSFRS_Delta'].diff()
df_subset['is_first_visit'] = df_subset['delta_t'].isna().astype(int)
df_subset['delta_t'] = df_subset['delta_t'].fillna(0) / 30.0  # Scale to months

# Prepare ordinal response matrix (0 for missing, 1-5 for actual scores)
Y_matrix = np.zeros((len(df_subset), 12), dtype=int)
for i, col in enumerate(items_cols):
    vals = df_subset[col].values
    Y_matrix[:, i] = np.where(np.isnan(vals), 0, vals + 1)

# Factor map based on Tran et al. 2021 mapping
factor_map = [1, 1, 1, 2, 2, 2, 3, 3, 3, 4, 4, 4]

stan_data = {
    'N': len(df_subset),
    'Nsub': df_subset['stan_subj_id'].nunique(),
    'K': 12,
    'D': 4,
    'ID': df_subset['stan_subj_id'].values.astype(int),
    'deltat': df_subset['delta_t'].values,
    'is_first_visit': df_subset['is_first_visit'].values.astype(int),
    'Y': Y_matrix,
    'factor_map': factor_map
}

print(f"Data Prepared: {stan_data['Nsub']} subjects, {stan_data['N']} total observations.")

# ---------------------------------------------------------
# 2. MODEL FITTING 
# ---------------------------------------------------------
print("Compiling Stan Model...")
model = CmdStanModel(stan_file='alsfrs_ou_4factor.stan')

print("Starting MCMC Sampling (show_console=True to see any C++ errors)...")
fit = model.sample(
    data=stan_data,
    chains=4,
    iter_warmup=300,
    iter_sampling=300,
    adapt_delta=0.90,
    max_treedepth=12,
    show_console=True # Shows progress and catches instant abort errors
)

# ---------------------------------------------------------
# 3. EXTRACTION & VISUALIZATION
# ---------------------------------------------------------
try:
    print("\nExtracting Factor Trajectories ('xi')...")
    
    # Extract 'xi' -> Shape is (Draws, N, D)
    xi_samples = fit.stan_variable('xi') 
    xi_mean = xi_samples.mean(axis=0)  # Average across draws -> Shape: (N, 4)
    
    df_subset['Latent_Bulbar'] = xi_mean[:, 0]
    df_subset['Latent_FineMotor'] = xi_mean[:, 1]
    df_subset['Latent_GrossMotor'] = xi_mean[:, 2]
    df_subset['Latent_Respiratory'] = xi_mean[:, 3]

    # Plot Latent Progression for 4 Random Patients
    sample_pats = np.random.choice(df_subset['stan_subj_id'].unique(), size=4, replace=False)
    fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=False, sharey=True)
    sns.set_theme(style="whitegrid")

    for idx, subj in enumerate(sample_pats):
        ax = axes.flatten()[idx]
        subj_data = df_subset[df_subset['stan_subj_id'] == subj]
        x_time = subj_data['ALSFRS_Delta'] / 30.0  # Months
        
        ax.plot(x_time, subj_data['Latent_Bulbar'], marker='o', label='Bulbar', color='#3498db')
        ax.plot(x_time, subj_data['Latent_FineMotor'], marker='s', label='Fine Motor', color='#2ecc71')
        ax.plot(x_time, subj_data['Latent_GrossMotor'], marker='^', label='Gross Motor', color='#f39c12')
        ax.plot(x_time, subj_data['Latent_Respiratory'], marker='d', label='Respiratory', color='#e74c3c')
        
        ax.set_title(f'Latent 4-Factor Progression (Patient {subj})', fontweight='bold')
        ax.set_xlabel('Time (Months from Baseline)')
        ax.set_ylabel('Latent Trait Score (Centered at 0)')
        if idx == 0:
            ax.legend()

    plt.tight_layout()
    plt.savefig("Latent_OU_Trajectories.png", dpi=300)
    print("Success! Saved trajectory plot to 'Latent_OU_Trajectories.png'")
    
    # Output the Latent Domain Correlation Matrix
    L_Omega = fit.stan_variable('L_Omega').mean(axis=0)
    Omega = np.dot(L_Omega, L_Omega.T)
    print("\nEstimated Latent Domain Correlation Matrix (Omega):")
    print(pd.DataFrame(Omega, 
                       columns=['Bulbar', 'FineMotor', 'GrossMotor', 'Respiratory'], 
                       index=['Bulbar', 'FineMotor', 'GrossMotor', 'Respiratory']).round(3))

except ValueError as e:
    print(f"\nExtraction Failed: {e}")
    print("This implies the model did not successfully complete sampling. Check the console output above for divergent transitions or rejection errors.")
```

```python
rlt_dir = '../PROACT_results/'
if not os.path.exists(rlt_dir):
    os.makedirs(rlt_dir)

# =========================================================
# 4. TABLE 2: Posterior Parameter Estimates
# =========================================================
print("\n--- TABLE 2: Posterior Parameter Estimates ---")
def format_estimate(samples):
    mean = np.mean(samples)
    sd = np.std(samples)
    lower = np.percentile(samples, 2.5)
    upper = np.percentile(samples, 97.5)
    return f"{mean:.3f} (SD: {sd:.3f}) [{lower:.3f}, {upper:.3f}]"

# Extract Loadings (Lambda)
lambda_samples = fit.stan_variable('lambda')
print("\nFactor Loadings (Lambda):")
for k, col in enumerate(items_cols):
    print(f"  Item {k+1} ({col}): {format_estimate(lambda_samples[:, k])}")

# Extract Mean Reversion Rates (Gamma)
gamma_samples = fit.stan_variable('gamma')
factors = ['Bulbar', 'Fine Motor', 'Gross Motor', 'Respiratory']
print("\nDrift / Reversion Rates (Gamma):")
for d, factor in enumerate(factors):
    print(f"  {factor}: {format_estimate(gamma_samples[:, d])}")

# Extract Correlation Matrix (Omega)
L_Omega_samples = fit.stan_variable('L_Omega')
# Reconstruct full Omega matrix for each draw
draws = L_Omega_samples.shape[0]
Omega_samples = np.array([np.dot(L_Omega_samples[i], L_Omega_samples[i].T) for i in range(draws)])

print("\nLatent Stationary Correlation Matrix (Omega) - Means:")
omega_mean = np.mean(Omega_samples, axis=0)
omega_df = pd.DataFrame(omega_mean, columns=factors, index=factors).round(3)
print(omega_df)

# =========================================================
# 5. FIGURE 2: Estimated Latent Trajectories with 95% CrI
# =========================================================
print("\nGenerating Figure 2 (Latent Trajectories with Uncertainty Bands)...")
xi_samples = fit.stan_variable('xi') # Shape: (Draws, N, D)

# Calculate Mean, 2.5%, and 97.5% percentiles for the latent states
xi_mean = np.mean(xi_samples, axis=0)
xi_lower = np.percentile(xi_samples, 2.5, axis=0)
xi_upper = np.percentile(xi_samples, 97.5, axis=0)

fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=False, sharey=True)
sns.despine()

# Pick 4 random patients to visualize
sample_pats_fig2 = np.random.choice(df_subset['stan_subj_id'].unique(), size=4, replace=False)
colors = ['#3498db', '#2ecc71', '#f39c12', '#e74c3c']

for idx, subj in enumerate(sample_pats_fig2):
    ax = axes.flatten()[idx]
    
    # Get row indices for this subject
    subj_mask = (df_subset['stan_subj_id'] == subj).values
    x_time = df_subset.loc[subj_mask, 'ALSFRS_Delta'] / 30.0
    
    for d, (factor, color) in enumerate(zip(factors, colors)):
        y_mean = xi_mean[subj_mask, d]
        y_lower = xi_lower[subj_mask, d]
        y_upper = xi_upper[subj_mask, d]
        
        # Plot mean line
        ax.plot(x_time, y_mean, marker='o', label=factor, color=color, linewidth=2)
        # Fill 95% Credible Interval
        ax.fill_between(x_time, y_lower, y_upper, color=color, alpha=0.15)
        
    ax.set_title(f'Latent Disease State: Patient {subj}', fontweight='bold')
    ax.set_xlabel('Time (Months from Baseline)')
    ax.set_ylabel('Latent Domain Score $\\xi(t)$')
    if idx == 0:
        ax.legend(loc='lower left')

plt.tight_layout()
plt.savefig(rlt_dir + "Figure_2_Latent_Trajectories_CrI.png", dpi=300)
plt.close()

print("\nSuccess! Paper tables printed to console. Figures saved to disk as PNG files.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from cmdstanpy import CmdStanModel
import os

# Set visualization style to match academic papers
sns.set_theme(style="white", context="paper", font_scale=1.2)

# =========================================================
# 1. DATA PREPARATION & TABLE 1 (Baseline Characteristics)
# =========================================================
DATA_DIR = '../data/PRO-ACT_Data/2026_02_27_PROACT_ALL_FORMS/'
print("Loading datasets...")
df_als = pd.read_csv(DATA_DIR + 'F_PROACT_ALSFRS.csv', low_memory=False)
df_demo = pd.read_csv(DATA_DIR + 'F_PROACT_DEMOGRAPHICS.csv', low_memory=False)
df_hist = pd.read_csv(DATA_DIR + 'F_PROACT_ALSHISTORY.csv', low_memory=False)

rlt_dir = '../PROACT_results/'
if not os.path.exists(rlt_dir):
    os.makedirs(rlt_dir)

# Define items
items_cols = [
    'Q1_Speech', 'Q2_Salivation', 'Q3_Swallowing',                 
    'Q4_Handwriting', 'Q5_Cutting', 'Q6_Dressing_and_Hygiene',     
    'Q7_Turning_in_Bed', 'Q8_Walking', 'Q9_Climbing_Stairs',       
    'R_1_Dyspnea', 'R_2_Orthopnea', 'R_3_Respiratory_Insufficiency'
]

# Clean ALSFRS data
df_als['Q5a_Cutting_without_Gastrostomy'] = pd.to_numeric(df_als['Q5a_Cutting_without_Gastrostomy'], errors='coerce')
df_als['Q5b_Cutting_with_Gastrostomy'] = pd.to_numeric(df_als['Q5b_Cutting_with_Gastrostomy'], errors='coerce')
df_als['Q5_Cutting'] = df_als['Q5a_Cutting_without_Gastrostomy'].fillna(df_als['Q5b_Cutting_with_Gastrostomy'])

for col in items_cols + ['ALSFRS_Delta']:
    if col in df_als.columns:
        df_als[col] = pd.to_numeric(df_als[col], errors='coerce')

df_als = df_als.dropna(subset=['ALSFRS_Delta', 'R_1_Dyspnea'])
df_als = df_als.sort_values(by=['subject_id', 'ALSFRS_Delta'])
df_als = df_als.drop_duplicates(subset=['subject_id', 'ALSFRS_Delta'], keep='first')

# Sample 100 subjects for tractable MCMC execution
np.random.seed(42)
sampled_subjects = np.random.choice(df_als['subject_id'].unique(), size=100, replace=False)
df_subset = df_als[df_als['subject_id'].isin(sampled_subjects)].copy()
df_subset['stan_subj_id'] = df_subset.groupby('subject_id').ngroup() + 1

# --- GENERATE TABLE 1: Baseline Characteristics ---
print("\n--- TABLE 1: Baseline Characteristics ---")
demo_sub = df_demo[df_demo['subject_id'].isin(sampled_subjects)].copy()
demo_sub['Age'] = pd.to_numeric(demo_sub['Age'], errors='coerce')
hist_sub = df_hist[df_hist['subject_id'].isin(sampled_subjects)].copy()

age_mean = demo_sub['Age'].mean()
age_sd = demo_sub['Age'].std()
sex_counts = demo_sub['Sex'].str.title().value_counts(normalize=True) * 100

valid_onset = hist_sub[hist_sub['Site_of_Onset'].str.contains('Onset', na=False)]
onset_counts = valid_onset['Site_of_Onset'].str.title().value_counts(normalize=True) * 100

print(f"Total Subjects (N): {len(sampled_subjects)}")
print(f"Total Observations: {len(df_subset)}")
print(f"Age at Baseline: {age_mean:.1f} (SD: {age_sd:.1f})")
print("Gender:")
for sex, pct in sex_counts.items():
    print(f"  {sex}: {pct:.1f}%")
print("Site of Onset:")
for site, pct in onset_counts.items():
    print(f"  {site}: {pct:.1f}%")

# =========================================================
# 2. FIGURE 1: Raw Empirical Trajectories
# =========================================================
print("\nGenerating Figure 1 (Raw Trajectories)...")
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)
sns.despine()

sample_pats_fig1 = np.random.choice(sampled_subjects, size=15, replace=False)
plot_data = df_subset[df_subset['subject_id'].isin(sample_pats_fig1)]

domains = [
    ('Bulbar (Q1 Speech)', 'Q1_Speech'),
    ('Fine Motor (Q4 Handwriting)', 'Q4_Handwriting'),
    ('Gross Motor (Q8 Walking)', 'Q8_Walking'),
    ('Respiratory (R1 Dyspnea)', 'R_1_Dyspnea')
]

for ax, (title, col) in zip(axes.flatten(), domains):
    for subj in sample_pats_fig1:
        pat_df = plot_data[plot_data['subject_id'] == subj]
        ax.plot(pat_df['ALSFRS_Delta'] / 30.0, pat_df[col], marker='o', markersize=3, alpha=0.5, linewidth=1)
    ax.set_title(title)
    ax.set_ylabel("Ordinal Score (0-4)")
    ax.set_yticks([0, 1, 2, 3, 4])

axes[1, 0].set_xlabel("Time (Months)")
axes[1, 1].set_xlabel("Time (Months)")
plt.tight_layout()
plt.savefig(rlt_dir + "Figure_1_Raw_Trajectories.png", dpi=300)
plt.close()

# =========================================================
# 3. STAN MODEL PREPARATION & FITTING
# =========================================================
df_subset['delta_t'] = df_subset.groupby('subject_id')['ALSFRS_Delta'].diff().fillna(0) / 30.0
df_subset['is_first_visit'] = (df_subset['delta_t'] == 0).astype(int)

Y_matrix = np.zeros((len(df_subset), 12), dtype=int)
for i, col in enumerate(items_cols):
    vals = df_subset[col].values
    Y_matrix[:, i] = np.where(np.isnan(vals), 0, vals + 1)

stan_data = {
    'N': len(df_subset),
    'Nsub': df_subset['stan_subj_id'].nunique(),
    'K': 12,
    'D': 4,
    'ID': df_subset['stan_subj_id'].values.astype(int),
    'deltat': df_subset['delta_t'].values,
    'is_first_visit': df_subset['is_first_visit'].values.astype(int),
    'Y': Y_matrix,
    'factor_map': [1, 1, 1, 2, 2, 2, 3, 3, 3, 4, 4, 4]
}

print("\nCompiling and Fitting Latent OU Model (Please wait...)")
model = CmdStanModel(stan_file='alsfrs_ou_4factor.stan')
fit = model.sample(
    data=stan_data,
    chains=4,
    iter_warmup=500,
    iter_sampling=500,
    adapt_delta=0.95, # High adapt_delta needed for SDEs
    max_treedepth=12,
    show_console=False
)

# =========================================================
# 4. MCMC DIAGNOSTICS & CONVERGENCE CHECKS
# =========================================================
print("\n=== MCMC CONVERGENCE DIAGNOSTICS ===")

# Run CmdStan's built-in diagnosis for Divergences, Tree Depth, and BFMI
print("\n--- HMC Sampler Health ---")
print(fit.diagnose())

# Extract parameter summary (contains R-hat and ESS)
fit_summary = fit.summary()

# 1. Check Gelman-Rubin R-hat statistic
r_hats = fit_summary['R_hat'].dropna()
bad_r_hats = r_hats[r_hats > 1.05]
print("\n--- R-hat (Potential Scale Reduction Factor) ---")
if len(bad_r_hats) > 0:
    print(f"Warning: {len(bad_r_hats)} parameters have R_hat > 1.05 (Poor Convergence).")
    print("Worst offenders:")
    print(bad_r_hats.sort_values(ascending=False).head(5))
else:
    print("Success: All parameters converged according to R_hat (< 1.05).")

# 2. Check Effective Sample Size (Bulk ESS)
ess_bulk = fit_summary['EssBulk'].dropna()
low_ess = ess_bulk[ess_bulk < 100]
print("\n--- Bulk Effective Sample Size (ESS) ---")
if len(low_ess) > 0:
    print(f"Warning: {len(low_ess)} parameters have extremely low Bulk-ESS (< 100).")
    print("Worst offenders:")
    print(low_ess.sort_values().head(5))
else:
    print("Success: All parameters have sufficient Effective Sample Size (ESS > 100).")

# 3. Plot R-hat Distribution
plt.figure(figsize=(8, 4))
sns.histplot(r_hats, bins=50, color='teal', kde=False)
plt.axvline(1.05, color='red', linestyle='--', label='Convergence Threshold (1.05)')
plt.title('Distribution of Gelman-Rubin $\hat{R}$ Statistic')
plt.xlabel('$\hat{R}$ (Values ~1.0 indicate convergence)')
plt.ylabel('Parameter Count')
plt.legend()
plt.tight_layout()
plt.savefig(rlt_dir + r"Diagnostic_Rhat_Distribution.png", dpi=300)
plt.close()

# =========================================================
# 5. TABLE 2: Posterior Parameter Estimates
# =========================================================
print("\n--- TABLE 2: Posterior Parameter Estimates ---")
def format_estimate(samples):
    mean = np.mean(samples)
    sd = np.std(samples)
    lower = np.percentile(samples, 2.5)
    upper = np.percentile(samples, 97.5)
    return f"{mean:.3f} (SD: {sd:.3f}) [{lower:.3f}, {upper:.3f}]"

# Extract Loadings (Lambda)
lambda_samples = fit.stan_variable('lambda')
print("\nFactor Loadings (Lambda):")
for k, col in enumerate(items_cols):
    print(f"  Item {k+1} ({col}): {format_estimate(lambda_samples[:, k])}")

# Extract Mean Reversion Rates (Gamma)
gamma_samples = fit.stan_variable('gamma')
factors = ['Bulbar', 'Fine Motor', 'Gross Motor', 'Respiratory']
print("\nDrift / Reversion Rates (Gamma):")
for d, factor in enumerate(factors):
    print(f"  {factor}: {format_estimate(gamma_samples[:, d])}")

# Extract Correlation Matrix (Omega)
L_Omega_samples = fit.stan_variable('L_Omega')
draws = L_Omega_samples.shape[0]
Omega_samples = np.array([np.dot(L_Omega_samples[i], L_Omega_samples[i].T) for i in range(draws)])

print("\nLatent Stationary Correlation Matrix (Omega) - Means:")
omega_mean = np.mean(Omega_samples, axis=0)
omega_df = pd.DataFrame(omega_mean, columns=factors, index=factors).round(3)
print(omega_df)

# =========================================================
# 6. FIGURE 2: Estimated Latent Trajectories with 95% CrI
# =========================================================
print("\nGenerating Figure 2 (Latent Trajectories with Uncertainty Bands)...")
xi_samples = fit.stan_variable('xi') # Shape: (Draws, N, D)

# Calculate Mean, 2.5%, and 97.5% percentiles for the latent states
xi_mean = np.mean(xi_samples, axis=0)
xi_lower = np.percentile(xi_samples, 2.5, axis=0)
xi_upper = np.percentile(xi_samples, 97.5, axis=0)

fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=False, sharey=True)
sns.despine()

sample_pats_fig2 = np.random.choice(df_subset['stan_subj_id'].unique(), size=4, replace=False)
colors = ['#3498db', '#2ecc71', '#f39c12', '#e74c3c']

for idx, subj in enumerate(sample_pats_fig2):
    ax = axes.flatten()[idx]
    
    subj_mask = (df_subset['stan_subj_id'] == subj).values
    x_time = df_subset.loc[subj_mask, 'ALSFRS_Delta'] / 30.0
    
    for d, (factor, color) in enumerate(zip(factors, colors)):
        y_mean = xi_mean[subj_mask, d]
        y_lower = xi_lower[subj_mask, d]
        y_upper = xi_upper[subj_mask, d]
        
        ax.plot(x_time, y_mean, marker='o', label=factor, color=color, linewidth=2)
        ax.fill_between(x_time, y_lower, y_upper, color=color, alpha=0.15)
        
    ax.set_title(f'Latent Disease State: Patient {subj}', fontweight='bold')
    ax.set_xlabel('Time (Months from Baseline)')
    ax.set_ylabel('Latent Domain Score $\\xi(t)$')
    if idx == 0:
        ax.legend(loc='lower left')

plt.tight_layout()
plt.savefig(rlt_dir + "Figure_2_Latent_Trajectories_CrI.png", dpi=300)
plt.close()

print("\nSuccess! Paper tables printed to console. Figures saved to disk as PNG files.")
```
