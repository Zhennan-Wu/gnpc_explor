```python
import numpy as np
from sklearn.decomposition import FactorAnalysis

# -----------------------
# fake data
# -----------------------
n_samples = 50000      # MUST be large
n_features = 7000
n_factors = 20

X = np.random.randn(n_samples, n_features)

# -----------------------
# Factor Analysis
# -----------------------
fa = FactorAnalysis(
    n_components=n_factors,
    random_state=0
)

Z = fa.fit_transform(X)   # shape: (n_samples, 20)

print("Latent representation shape:", Z.shape)
```
